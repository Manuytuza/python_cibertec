# Matrices Numpy

Las matrices NumPy son la principal forma en que usaremos Numpy a lo largo del curso. Los arreglos Numpy vienen esencialmente en dos sabores: vectores y matrices. # Indexación y selección de NumPy

En esta lección discutiremos cómo seleccionar elementos o grupos de elementos de una matriz.

```python
arr = np.arange(0,11)
arr
```


## Indexación y selección de soportes
La forma más sencilla de elegir uno o algunos elementos de una matriz es muy similar a las listas de Python:

```python
#Obtener un valor en un índice
arr[8]
```

```python
#Obtener valores en un rango
arr[1: 5]
```

## Broadcasting
Las matrices Numpy difieren de una lista de Python normal debido a su capacidad para transmitir:

```python
#Configuración de un valor con rango de índice (difusión)
arr[0:5] = 100
arr
```

```python
# Reset array, veremos por qué tuve que reiniciar en un momento
arr = np.arange(0,11)
arr
```

```python
#Notas importantes sobre Slices
slice_of_arr = arr[0: 6]
slice_of_arr
```

```python
#Cambio Slice
slice_of_arr[:]=99
slice_of_arr
```

¡Ahora tenga en cuenta que los cambios también ocurren en nuestra matriz original!

```python
arr
```

```python
# Para obtener una copia, debe ser explícita.
arr_copy = arr.copy()

arr_copy
```


## Indexación de una matriz 2D (matrices)

El formato general es ** arr_2d [fila] [col] ** o ** arr_2d [fila, col] **. Por lo general, recomiendo usar la notación de coma para mayor claridad.

```python
arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))
arr_2d
```

```python
# Fila de indexación
arr_2d[1]
```

```python
# El formato es arr_2d [fila] [col] o arr_2d [fila, col]

# Obtener valor de elemento individual
arr_2d[1][0]
```

```python
# Obtener valor de elemento individual
arr_2d [1,0]
```

```python
#Slicing de matriz 2D

#Shape (2,2) desde la esquina superior derecha
arr_2d [: 2,1:]
```

```python
#Shape fila inferior
arr_2d [2]
```

### Indexación elegante

La indexación elegante le permite seleccionar filas o columnas enteras fuera de orden, para mostrar esto, construyamos rápidamente una matriz numpy:

```python
#Configurar matriz
arr2d = np.zeros((10,10))
```

```python
arr_length = arr2d.shape[1]
```

```python
arr_length
```

```python
#Configurar matriz

for i in range(arr_length):
    arr2d[i] = i
    
arr2d
```

```python
arr2d[[2,4,6,8]]
```

```python
arr2d[[6,4,2,7]]
```

## Funciones de matriz universal

Numpy viene con muchas [funciones de matriz universales] (http://docs.scipy.org/doc/numpy/reference/ufuncs.html), que son esencialmente operaciones matemáticas que puede utilizar para realizar la operación en toda la matriz. Muestremos algunos comunes:

```python
# Tomando raíces cuadradas
np.sqrt(arr)
```

```python
#Calcular exponencial (e^)
np.exp (arr)
```

```python
np.max(arr) #igual que arr.max()
```

```python
np.sin(arr)
```

```python
np.log(arr)
```
