```python
import pandas as pd

def clasificar_nota(nota):
    if nota>=13:
        return"Aprobado"
    else:
        return"Desaprobado"

def pedir_datos(maximos_estudiantes=5):
    nombres=[]
    edades=[]
    notas=[]

    print("Ingresa los datos de los estudiantes.")
    print("Escribe 'salir' como nombre para terminar antes de tiempo.\n")

    for i in range(maximos_estudiantes):
        try:
            nombre=input(f"Nombre del estudiante {i + 1}: ").strip()
            if nombre.lower()=="salir":
                print("Ingreso de datos finalizado por el usuario.\n")
                break

            if nombre=="":
                print("Nombre vacío. Se omite este registro.\n")
                continue

            edad=int(input("Edad: "))
            nota=float(input("Nota final (0 a 20): "))

            if nota<0 or nota>20:
                print("La nota debe estar entre 0 y 20. Registro omitido.\n")
                continue

            nombres.append(nombre)
            edades.append(edad)
            notas.append(nota)

            print("Registro guardado.\n")

        except ValueError:
            print("Error: edad y nota deben ser números válidos.\n")
            continue

    return nombres,edades,notas

def lista_de_alumnos(nombres,edades,notas):
    df=pd.DataFrame({
        "Nombre": nombres,
        "Edad": edades,
        "Nota": notas
    })

    df["Estado"]=df["Nota"].apply(clasificar_nota)
    return df

def mostrar_resumen(df):
    if df.empty:
        print("No se registraron estudiantes.")
    else:
        print("\n         Lista de estudiantes")
        print(df)

        print("\n         RESUMEN")
        print("Total de estudiantes:", len(df))
        print("Promedio de notas:", round(df["Nota"].mean(), 2))
        print("Aprobados:", (df["Estado"]=="Aprobado").sum())
        print("Desaprobados:", (df["Estado"]=="Desaprobado").sum())

        if df["Nota"].mean() >= 13:
            print("Promedio general: bueno.")
        else:
            print("Promedio general: necesita mejorar.")

def main():
    nombres,edades,notas=pedir_datos(5)
    df=lista_de_alumnos(nombres,edades,notas)
    mostrar_resumen(df)

if __name__ == "__main__":
    main()
```
