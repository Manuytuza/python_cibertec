# Tarea - Listas

Un profesor necesita registrar las notas de su último examen de matemáticas. La dirección le ha pedido enviarle un archivo en csv de aquellos alumnos que obtuvieron menos o igual a 11, para implementar tutorías por las tardes.

```python
import pandas as pd

alumnos = []

while True:
    nombre = input("Nombre del alumno: ")
    nota = float(input("Nota: "))

    alumno = {
        "nombre": nombre,
        "nota": nota
    }

    alumnos.append(alumno)

    continuar = input("¿Agregar otro alumno? (s/n): ")
    if continuar.lower() != "s":
        break
```

```python
df = pd.DataFrame(alumnos)
print(df)
```

```python
# Filtrar alumnos con nota menor a 11
df_tutoria = df[df["nota"] <= 11]
print(df_tutoria)
```

```python
# Guardar lista en CSV
df_tutoria.to_csv("alumnos_tutoria.csv", index=False)
```
